## Lab 3 - Tax Functions

This lab demonstrates creating custom functions to calculate Social Security tax, Medicare tax, and Federal tax withholding based on gross pay and withholding codes.

```python
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062
```

```python
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145
```

```python
def get_federal_tax(gross_pay, withholding_code):

    if withholding_code == 0:
        rate = 0.23
    elif withholding_code == 1:
        rate = 0.21
    elif withholding_code == 2:
        rate = 0.195
    elif withholding_code == 3:
        rate = 0.185
    else:
        rate = 0.18 - ((withholding_code - 4) * 0.005)

    return gross_pay * rate
```

```python
gross_pay = 750
code = 0

print("Person 1")
print("Social Security Tax:", get_soc_sec_tax(gross_pay))
print("Medicare Tax:", get_medicare_tax(gross_pay))
print("Federal Tax:", get_federal_tax(gross_pay, code))
```
