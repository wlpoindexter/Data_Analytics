# Exercise 2.C Creating and calling functions
## Lab 1. Simple functions

```python
def favorite_things(name, movie, music):
    return f"Name: {name}\nFavorite movie: {movie}\nFavorite music: {music}"
```

```python
print(favorite_things("Will", "The Dark Knight", "Drake"))
```

```python
print(favorite_things("Will", "The Dark Knight", "Drake"))
```

```python
def why_im_here(purpose):
    return purpose
```

```python
print(why_im_here("I joined the Year Up United Data Analyst program to build stronger technical skills, gain professional experience, and start a career in data analytics."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("Chicago", "there is always something to do and it feels like home"))
```

```python
print(favorite_place("the gym", "it helps me stay focused and keep improving myself"))
```
