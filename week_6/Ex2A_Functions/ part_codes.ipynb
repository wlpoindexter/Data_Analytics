## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(code):
    parts = code.split(":")
    return parts[0]

get_supplier_code("ACE:1-12")

def get_product_name(code):
    parts = code.split(":")
    print(parts)
    
    print (parts[1])
    p = parts[1].split("-")
    return (p[0])

get_product_name("ACE:1-12")

def get_size(code):
    parts = code.split("-")
    return(parts[1])

get_size("ACE:1-12")
```
