## Lab 1. Working with names – parsing strings

```python
test_name = "John Doe"

print(test_name[:4])
print(test_name[5:])
```

```python
def parse_and_display(name):
    print(name)
    for i in range(len(name)):
        print(i)
        if name[i] == ' ':
            first = name[:i]
            last = name[i+1:]
            print(f"Full name: {name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")

parse_and_display(test_name)
```

```python
test_list = ['Grace Flores', 'JB Oinonen', 'Ken Penn']
```

```python
parse_and_display(test_list[1])
print(test_list[1])
```

```python
test_list = "Will Poindexter"
```

```python
parse_and_display(test_list)
```

```python
test_list = ["Lorde", "Billie Eilish", "Megan Thee Stallion"]

def parse_and_display2(name):
    parts = name.split()

    print(f"Full name: {name}")

    if len(parts) == 1:
        print(f"First name: {parts[0]}")
    elif len(parts) == 2:
        print(f"First name: {parts[0]}")
        print(f"Last name: {parts[1]}")
    elif len(parts) == 3:
        print(f"First name: {parts[0]}")
        print(f"Middle name: {parts[1]}")
        print(f"Last name: {parts[2]}")

    print()

parse_and_display2(test_list[0])
parse_and_display2(test_list[1])
parse_and_display2(test_list[2])
```
